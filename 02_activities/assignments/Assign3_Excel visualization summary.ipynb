EXCEL VISUALIZATION: Outbreaks by Healthcare Setting in Toronto (2026)

Link to Outbreaks in Toronto Healthcare Institutions data: https://ckan0.cf.opendata.inter.prod-toronto.ca/dataset/80ce0bd7-adb2-4568-b9d7-712f6ba38e4e/resource/be82644f-a9d8-449a-badb-c138fe5b83b6/download/ob_report_2026.csv

> What software did you use to create your data visualization?

In Microsoft Excel, PivotTable and PivotChart functions were used to create and summarize the outbreak data by healthcare setting.

> Who is your intended audience?

The intended audience includes public health professionals, healthcare administrators, students, and members of the public who are interested in understanding patterns of infectious disease outbreaks in healthcare settings.
    
> What information or message are you trying to convey with your visualization? 

This visualization shows how outbreaks are distributed across different healthcare settings in Toronto in 2026, with outbreak types represented by color coding. It highlights that long-term care homes (LTCHs) experienced substantially more outbreaks than hospitals or retirement homes, and that respiratory outbreaks were the most common type across all settings.
    
> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 

I prioritized simplicity and readability when designing this visualization. A bar chart was chosen because it allows easy comparison across healthcare settings. Clear axis labels and a descriptive title were included to provide context and help viewers interpret the data. Distinct colors were used to differentiate outbreak types and allow viewers to compare patterns across healthcare settings. The colors were chosen to be visually distinct while avoiding unnecessary complexity. A legend was included so that information was not conveyed by color alone, improving the accessibility of the visualization.
    
> How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization?

The visualization can be reproduced by using the same dataset and following the documented PivotTable and PivotChart steps in Excel. However, Excel is less reproducible than code-based tools because some formatting and manual steps may not be automatically recorded. This may make exact replication more difficult compared with Python scripts, which can reproduce the same visualization automatically.
    
> How did you ensure that your data visualization is accessible? 

Accessibility was ensured by using clear axis labels, readable font sizes, and a descriptive title. Distinct colors and a legend were used to differentiate outbreak types, ensuring that viewers could interpret the chart without relying solely on color. The simple layout also makes comparisons between healthcare settings easy to understand. 
    
> Who are the individuals and communities who might be impacted by your visualization?  

This visualization may be relevant to long-term care residents, retirement home residents, patients, healthcare workers, public health agencies, and policymakers responsible for infection prevention and control. The findings may also be of interest to family members and the general public concerned about healthcare-associated outbreaks.

> How did you choose which features of your chosen dataset to include or exclude from your visualization? 

I included healthcare setting, type of outbreak, and outbreak counts because these variables directly address how outbreaks differ across healthcare environments which was my goal. Variables such as institution name, address, and outbreak dates were excluded to keep the visualization focused and easy to interpret.

> What ‘underwater labour’ contributed to your final data visualization product?

Underwater labour included downloading the dataset, reviewing variable definitions, cleaning and organizing the data, creating PivotTables and PivotCharts, selecting an appropriate chart type, choosing colors and labels, and refining the visualization to improve readability and accessibility.
