```python
import pandas as pd
#Loading and inspecting a dataset in Python using pandas
df = pd.read_csv("ob_report_2026.csv")
print(df.head())
```

#Which pathogens were most frequently responsible for outbreaks?

```python
#counting frequency of each causative agent in the dataset, selecting 10 most common agents
top_agents = df['Causative Agent-1'].value_counts().head(10)
print(top_agents)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))  # creates a new figure with specified width and height

ax = top_agents.sort_values().plot(kind='barh') # sorts values and creates a horizontal bar chart

plt.title('Top 10 Causes of Healthcare Institution Outbreaks') # adds a title to the chart
plt.xlabel('Number of Outbreaks') # labels the x-axis
plt.ylabel('Causative Agent') # labels the y-axis

# add value labels
for i, v in enumerate(top_agents.sort_values()):
    ax.text(v + 0.5, i, str(v), va='center') # loops through values and adds numeric labels to each bar

plt.tight_layout() # adjusts layout to prevent overlapping elements
plt.savefig("top_agents.png", dpi=300, bbox_inches="tight") # saves the figure as a high-resolution PNG file
plt.show()
```
