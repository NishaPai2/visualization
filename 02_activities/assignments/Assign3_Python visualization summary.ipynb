PYTHON VISUALIZATION: Top 10 Causes of Healthcare Institution Outbreaks

Link to Outbreaks in Toronto Healthcare Institutions data: https://ckan0.cf.opendata.inter.prod-toronto.ca/dataset/80ce0bd7-adb2-4568-b9d7-712f6ba38e4e/resource/be82644f-a9d8-449a-badb-c138fe5b83b6/download/ob_report_2026.csv

> What software did you use to create your data visualization?
    
I used Python to create this visualization, specifically the pandas library for data manipulation and matplotlib for data visualization. The plot was generated within a Jupyter Notebook environment.

> Who is your intended audience? 
    
The intended audience includes public health professionals, healthcare workers, researchers, students, and members of the public interested in understanding the causes of outbreaks in healthcare institutions.

> What information or message are you trying to convey with your visualization? 

This visualization shows the most common causative agents responsible for outbreaks in Toronto healthcare institutions during 2026. The chart highlights which pathogens are most frequently associated with outbreaks and may help identify priorities for infection prevention and control.

> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 

I prioritized clarity and readability when designing the visualization. A horizontal bar chart was chosen because it allows long pathogen names to be displayed clearly and makes comparisons between categories easier. The causative agents were sorted by outbreak frequency to emphasize the most common pathogens. Axis labels, a descriptive title, and data labels displaying outbreak counts were added to improve interpretation of the visualization.
    
> How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization? 

The visualization is highly reproducible because all data cleaning, analysis, and plotting steps were performed using Python code. Running the same script on the same dataset will automatically reproduce the visualization. This improves transparency and reduces the possibility of manual errors.

    
> How did you ensure that your data visualization is accessible?  

Accessibility was improved by using clear axis labels, readable font sizes, and a descriptive title. The horizontal layout allows long pathogen names to be displayed without overlapping. Data labels were added directly to the bars so viewers can easily interpret the exact outbreak counts.

> Who are the individuals and communities who might be impacted by your visualization?  

This visualization may be relevant to patients, residents of long-term care homes and retirement homes, healthcare workers, infection prevention specialists, public health agencies, policymakers, and the general public.

> How did you choose which features of your chosen dataset to include or exclude from your visualization? 

I included the causative agent variable because it directly addresses which pathogens are responsible for outbreaks, which was my goal. To improve readability, I focused on the top ten most frequent causative agents and excluded variables such as institution names, addresses, and dates that were not directly relevant to my goal.

> What ‘underwater labour’ contributed to your final data visualization product?

Underwater labour included downloading and exploring the dataset, cleaning and organizing the data, counting outbreak frequencies, selecting an appropriate visualization type, writing and debugging Python code, adding labels and annotations, and refining the figure to improve readability and accessibility.
