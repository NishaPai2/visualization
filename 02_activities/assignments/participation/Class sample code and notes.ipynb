```python
#Mirror Histogram Chart from: https://python-graph-gallery.com/density-mirror/

# libraries
import numpy as np
from numpy import linspace
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# dataframe
df = pd.DataFrame({
'var1': np.random.normal(size=1000),
'var2': np.random.normal(loc=2, size=1000) * -1
})

# Fig size
plt.rcParams["figure.figsize"]=12,8

# plot histogram chart for var1
sns.histplot(x=df.var1, stat="density", bins=20, edgecolor='black')

# plot histogram chart for var2
n_bins = 20
# get positions and heights of bars
heights, bins = np.histogram(df.var2, density=True, bins=n_bins) 
# multiply by -1 to reverse it
heights *= -1
bin_width = np.diff(bins)[0]
bin_pos =( bins[:-1] + bin_width / 2) * -1

# plot
plt.bar(bin_pos, heights, width=bin_width, edgecolor='black')

# show the graph
plt.show()
```

# NOTES: Evaluating above Mirror Histogram Chart
Aesthetic qualitites: It is visually appealing and balanced; the contrasting blue and orange colors make the two distributions easy to distinguish.; mirrored layout is interesting and draws attention.
However, no axis labels, title, or legend is included, therefore it is difficult to know what the data represent; the mirrored format is unfamiliar to me and slightly confusion for direct comparison vs it being side-by-side  

Substantive qualities: the visualization effectively compares the shapes, spread, and frequencies of two distributions; allows viewers to quickly see differences in center, variability, and skewness between the groups.
However, without context or labels, I can't interpret the meaning of the distributions; exact values are difficult to extract. 

Perceptual qualities: Since many (including me) are good at comparing bar lengths, the histogram bins are easy to compare; the mirror design makes similarities and differences between distributions stand out.
However, because one histogram is inverted, some people may initially interpret the negative bars as negative values rather than simply a visual comparison technique; comparing exact frequencies across the mirrored axis may require more effort than with side-by-side histograms.

Overall, this mirror histogram does a good job visually because it's clean and easy to look at. It also works well for comparing two distributions, and the differences between them are fairly easy to spot. However, since there are no labels and one histogram is flipped upside down, I found it confusing and had trouble in interpreting it (i.e. understanding exactly what the graph is showing)

```python
#Multiple viz on one axes
#call multiple plot methods on the same axes
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

#now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
       color = "indigo")
ax.plot(x, y2,
        color = "red")
```

Note: Annotations, shapes, etc. can be added as usual.

```python
#Add error information
#First, calculate standard deviation of our data
y2_sd = np.std(y2)

#Then plot our line as before
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

#Then use errorbar() to add in our error line (standard deviation in this case, but could be whatever value you calculated)
ax.errorbar(x, #our x values 
            y2, #our y values 
            yerr = y2_sd, #yerr specifies that we’re plotting vertical error bars
            fmt = "none") #fmt makes sure we’re not plotting the actual data points, only the error
```

```python
#Customizing errorbar appearance
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, 
            y2, 
            yerr = y2_sd, 
            fmt = "none", 
            ecolor= "indigo", 
            elinewidth= 4, 
            capsize = 6, 
            capthick= 4 )
```

```python
#Errorevery
#When we don’t want to see error bars for every single point, specify intervals using errorevery
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, 
            y2, 
            yerr = y2_sd, 
            fmt = "none", 
            ecolor = "indigo", 
            elinewidth = 4, 
            capsize = 6, 
            capthick = 4, 
            errorevery= 2 )
```
